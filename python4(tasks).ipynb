```python
class Animal:
    def eat(self):
        print("Animal eats food")
class Bird(Animal):
    def fly(self):
         print("Bird can fly")
class Parrot(Bird):
    def speak(self):
        print("Parrot can speak")
c = Parrot()
c.eat()
c.fly()
c.speak()
```

```python
class animal:
    def __init__(self,name):
        self.name = name
    def eat(self):
        print("Animal eats food")
        print(f'{self.name} can swim')

class bird(animal):
    def __init__(self,name,name1):
        super().__init__(name)
        self.name1 = name1
    def fly(self):
        print("All Birds can fly")
        print(f'{self.name1} cant fly')

class parrot(bird):
    def __init__(self,name,name1):
        super().__init__(name,name1)
    def speak(self):
        self.eat()
        self.fly()
        print("parrot can speak")

s = parrot('fish','peacock')

s.speak()
```

```python
class Camera:
    def take_photo(self):
        print("taking photo")
class MusicPlayer:
    def play_music(self):
        print("playing music")
class Smartphone(Camera, MusicPlayer):
    def call(self):
        print("calling")
c = Smartphone()
c.take_photo()
c.play_music()
c.call()

```

```python
class person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def display_name(self):
        print(f"Name:{self.name}")
        print(f"Age:{self.age}")
class Employe(person):
    def __init__(self,name,age,salary):
        super().__init__(name,age)
        self.salary= salary
    def display_salary(self):
        self.display_name()
        print(f"Salary:{self.salary}")
s = Employe("Pavi",84,34000)
s.display_salary()
    
    
```

```python
class Wallet:
    def __init__(self,money):
        self.__money = money
    def deposit(self,amount):
        if amount > 0:
            self.__money += amount
    def show_money(self):
        print(f'Money Available:{self.__money}')
w = Wallet(1000)
w.show_money()
w.deposit(500)
w.show_money()
```

```python
#task 8 :
class ATM:
    def __init__(self,money):
        self.__money = money

    def withdraw(self,amount):
        if amount > self.__money:
            print("Insufficient Balance")
        else:
            self.__money -= amount
            print("withdraw succesful")
    def show_money(self):
        print(f'Remaining Money :{self.__money}')
w = ATM(10000)

w.withdraw(20000)
w.show_money()
```
