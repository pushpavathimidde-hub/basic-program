```python
class using parent class
```

```python
class father:
     def display(self):
         print("this is a parent class")

class mother(father):
    def show(self):
        print("this is a child class")
        
class child(mother):
     def show1(self):
         print("this is multiple inheritance")
c = child()
c.display()
c.show()
c.show1()
```

```python
class product:
    def __init__(self,name,price):
         self.name = name
         self.price = price
    
    def display(self):
        print(f"product:{self.name}")
        print(f"price:{self.price}")
class clothing(product):
    def __init__(self,name,price,warrenty):
       super().__init__(name,price)
       self.warrenty = warrenty

    def display1(self):
       self.display()
       print(f'warrenty:{self.warrenty}years')
c = clothing('shirt',2000,1)
c.display1()

                  
```

multilevel inheritance

```python
class emp:
    def __init__(self,id,name):
        self.id = id
        self.name = name
    def display(self):
        print(f"employee id:{self.id},\nEmployee NAME:{self.name}")
class pgm(emp):
    def __init__(self,id,name,language):
      super().__init__(id,name)
      self.language=language
    def display2(self):
        self.display()
        print(f"programming language:{self.language}")
class size(pgm):
    def __init__(self,id,name,language,teamsize):
        super().__init__(id,name,language)
        self.teamsize = teamsize
    def display3(self):
       self.display2()
       print(f"teamsize:{self.teamsize}")
c = size(12,'sandhya','python',3)
c.display3()

            

```

```python
class phone:
    def __init__(self,brand):
        self.brand = brand
    def display(self):
        print(f'Brand: {self.brand}')
class image:
    def show(self):
        print(f'capturing image with {self.brand} mobile')
class calling(phone, image):
    def __init__(self, brand):
        phone.__init__(self,brand)
    def say(self):
        self.display()
        self.show()
        print(f'calling with: {self.brand}')
c = calling('Iphone')
c.say()
```

```python
class bank:
    def __init__(self,name,balance):
        self.name = name
        self.__balance = balance

    def deposit(self,amount):
        if amount >0:
            self.__balance += amount
    def withdraw(self,amount):
        if amount > self.__balance:
            print("insufficient balance")
        else:
            self.__balance -= amount
    def get_balance(self):
        return self.__balance

acc = bank("name",1000)
acc.deposit(5000)
acc.withdraw(1000)
print(acc.get_balance())
```
