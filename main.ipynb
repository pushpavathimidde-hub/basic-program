```python
import pygame

pygame.init()

screen = pygame.display.set_mode((500, 500))
pygame.display.set_caption("My Game")

running = True

while running:

    screen.fill((255, 255, 255))        

    pygame.draw.circle(screen, (255, 0, 0), (250, 250), 50)

    pygame.display.update()               

    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

pygame.quit()
```

```python
#custom Iterator
class count:
    def __init__(self,start):
        self.current = start
    def __iter__(self):
        return self
    def __next__(self):
        if self.current == -1:
           raise StopIteration 
        val = self.current 
        self.current -= 1
        return val
for n in count(5):
    print(n)
    
```

```python
#Generators - a special function
#yield - it resume a fuction where it was stopped
#generattor - print only one value for one print statement
def fruits():
    yield "Apple"
    yield "Orange"
    yield "Grape"
    yield "Banana"
f = fruits()
print(next(f))
print(next(f))
print(next(f))
```

```python
#generator - using forloop instead of print statement
def fruits():
    yield "Apple"
    yield "Orange"
    yield "Grape"
    yield "Banana"
for fruit in fruits():
    print(fruit)
```

```python
#printing even numbers using generators
def even(n):
    for i in range(2,n+1,2):
        yield i
for num in even(10):
    print(num,end = " ")
```

```python
#printing power of even numbers
def even(n):
    for i in range(2,n+1,2):
        yield i**2
for num in even(10):
    print(num,end =" ")
```

```python
#printing power of 1-10 numbers
def even(n):
    for i in range(0,n+1,1):
        yield i**2
for num in even(10):
    print(num,end=" ")
```

```python
#Fibonacci series using generators
def even(n):
    a = 0
    b = 1
    for i in range(n):
        yield a
        a,b = b, a+b
for num in even(10):
    print(num,end=" ")
```

```python
#Decorators
def decarator_function(n):
    def wrapper():
        print("before function")
        n()
        print("after function")
    return wrapper
@decarator_function
def greet():
    print("Hello")
decorated = decarator_function(greet)
print(decorated())
```

```python
#Decorators - (way2) with arguments
def decarator_function(n):
    def wrapper(name):
        print("before function")
        n(name)
        print("after function")
    return wrapper
@decarator_function
def greet(name):
    print("Hello",name)
greet("Hello")
    
```
